```python
# Create a copy for preprocessing
df = asd_psychosis.copy()

print("Working dataset shape:", df.shape)
```

```python
# Separate input features (X) and target (y)

X = df.drop("risk_label", axis=1)
y = df["risk_label"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget classes:")
print(y.value_counts().sort_index())
```

```python
binary_features = [
    "gender",
    "speech_delay",
    "attention_deficit",
    "family_history_asd",
    "family_history_psychosis",
    "developmental_delay",
    "functional_decline"
]

score_features = [
    "language_score",
    "social_score",
    "repetitive_behavior",
    "cognitive_score",
    "anxiety_level",
    "emotional_regulation",
    "sleep_quality",
    "paranoia_traits",
    "perceptual_abnormalities"
]

age_feature = ["age"]

print("Binary features:", len(binary_features))
print("Score features:", len(score_features))
print("Age feature:", age_feature)
print("Total:", len(binary_features) + len(score_features) + len(age_feature))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set :", X_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting class distribution:")
print(y_test.value_counts().sort_index())
```

```python
print("========== DATA SPLIT SUMMARY ==========")

split_summary = pd.DataFrame({
    "Dataset": ["Original", "Training", "Testing"],
    "Samples": [len(y), len(y_train), len(y_test)],
    "Class_0": [
        (y == 0).sum(),
        (y_train == 0).sum(),
        (y_test == 0).sum()
    ],
    "Class_1": [
        (y == 1).sum(),
        (y_train == 1).sum(),
        (y_test == 1).sum()
    ],
    "Class_2": [
        (y == 2).sum(),
        (y_train == 2).sum(),
        (y_test == 2).sum()
    ]
})

display(split_summary)
```

```python
split_summary.to_csv(
    REPORTS_DIR / "train_test_split_summary.csv",
    index=False
)

print("Split summary saved successfully!")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Features that need scaling
numeric_features = ["age"] + score_features

# Binary features are already 0/1, so no scaling is required
preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", StandardScaler(), numeric_features),
        ("binary", "passthrough", binary_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape :", X_test_processed.shape)
```
